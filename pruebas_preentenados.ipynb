```python
from ultralytics import YOLO
import os

def entrenar_trigo_rapido():
    # Usamos el modelo más pequeño (Nano) para que vuele
    model = YOLO('yolov8n.pt') 

    path_yaml = os.path.join(os.getcwd(), 'GWHD2021-1', 'data.yaml')

    # CONFIGURACIÓN PARA 8-9 HORAS
    results = model.train(
        data=path_yaml,
        epochs=15,           # Bajamos de 30 a 15 (suficiente con 6000 fotos)
        imgsz=512,           # Bajamos de 640 a 512 (esto acelera un 30% el proceso)
        batch=16,            # Mantenemos 16 para aprovechar la RAM
        fraction=0.5,        # ¡CLAVE! Solo usamos el 50% del dataset (3000 fotos elegidas al azar)
        cache=True,          # Guardamos las fotos en RAM para no leer del disco lento
        workers=8,           # Usa todos los hilos de tu i7 al máximo
        name='wheat_fast_v1',
        device='cpu'
    )

if __name__ == '__main__':
    entrenar_trigo_rapido()
```
