```python
import requests

def obtener_ph_suelo(lat, lon):
    # Usamos la propiedad phh2o (pH en agua) para la capa de 0-5cm
    url = f"https://rest.isric.org/soilgrids/v2.0/properties/query?lon={lon}&lat={lat}&property=phh2o&depth=0-5cm&value=mean"
    
    try:
        response = requests.get(url, timeout=10)
        data = response.json()
        
        # Acceso seguro a la estructura del JSON
        layers = data.get('properties', {}).get('layers', [])
        if not layers:
            print(f"❌ No hay datos de suelo para las coordenadas: {lat}, {lon}")
            return

        valor_raw = layers[0]['depths'][0]['values']['mean']
        
        if valor_raw is not None:
            ph_valor = valor_raw / 10
            print(f"📍 Ubicación: Lat {lat}, Lon {lon}")
            print(f"🧪 pH estimado del suelo: {ph_valor}")
        else:
            print(f"⚠️ El punto {lat}, {lon} existe pero el valor de pH es nulo (posible zona urbana o agua).")

    except Exception as e:
        print(f"Error en la conexión: {e}")

# Ejemplo en una zona de campo (Cerca de Madrid, pero zona rural)
obtener_ph_suelo(lat=-33.19512257289546, lon=-70.72926807455951
)
```
